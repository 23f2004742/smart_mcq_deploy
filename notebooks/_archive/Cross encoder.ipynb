# Smart MCQ Solver

Predicts the **top 3 correct answers** for each MCQ using a 7-model soft score ensemble.

### Models
| # | Model | Notes |
|---|-------|-------|
| 1 | TF-IDF Cosine | n-gram lexical baseline |
| 2 | BM25 | probabilistic retrieval, complements TF-IDF |
| 3 | Sentence Transformer | `BAAI/bge-large-en-v1.5` |
| 4 | NLI DeBERTa-large | custom hypothesis template |
| 5 | RAG | ST-based retrieval + context reranking |
| 6 | Cross-Encoder | `ms-marco-MiniLM-L-12-v2`, direct pair scoring |
| 7 | MLP | trained on TF-IDF + ST + BM25 features (25-dim) |

**Ensemble**: softmax-normalized weighted average, weights tuned on val MAP@3.

**Metric**: MAP@3

## 4. Text Preprocessing

```python
def clean_text(text: str) -> str:
    """Clean prompts and options."""

    text = str(text).strip()
    text = re.sub(r"\s+", " ", text)

    boilerplate = [
        r"^Pick the best possible answer:\s*",
        r"^Choose the correct answer:\s*",
        r"^Select the most accurate option:\s*",
        r"^Identify the correct statement:\s*",
        r"^Determine the correct option:\s*",
        r"\s*among the listed options\.?$",
        r"\s*from the following choices\.?$",
        r"\s*based on the given context\.?$",
        r"\s*carefully\.?$",
    ]

    for pattern in boilerplate:
        text = re.sub(pattern, "", text, flags=re.IGNORECASE).strip()

    return text


def preprocess_row(row):
    """Return a cleaned question."""

    return {
        "prompt": clean_text(row["prompt"]),
        "A": clean_text(row["A"]),
        "B": clean_text(row["B"]),
        "C": clean_text(row["C"]),
        "D": clean_text(row["D"]),
        "E": clean_text(row["E"]),
    }


# Apply preprocessing
train_split = train_split.copy()
val_split = val_split.copy()
test_df_clean = test_df.copy()

for df in [train_split, val_split, test_df_clean]:
    df["prompt"] = df["prompt"].apply(clean_text)

    for opt in OPTIONS:
        df[opt] = df[opt].apply(clean_text)

print("Text preprocessing completed.")

raw = train_df.iloc[0]["prompt"]
clean = train_split.iloc[0]["prompt"]

print("\nBefore:", raw[:100])
print("After :", clean[:100])
```

## 5. Evaluation Metric (MAP@3)

```python
def average_precision_at_3(actual, predicted):
    """Compute AP@3 for one prediction."""

    predicted = [str(p).strip().upper() for p in predicted[:3]]

    for rank, pred in enumerate(predicted, start=1):
        if pred == actual:
            return 1 / rank

    return 0.0


def map_at_3(actuals, predictions):
    """Compute Mean Average Precision@3."""

    assert len(actuals) == len(predictions)

    scores = [
        average_precision_at_3(a, p)
        for a, p in zip(actuals, predictions)
    ]

    return float(np.mean(scores))


# Quick verification
assert average_precision_at_3("A", ["A", "B", "C"]) == 1
assert average_precision_at_3("A", ["B", "A", "C"]) == 0.5
assert abs(average_precision_at_3("A", ["C", "D", "A"]) - 1 / 3) < 1e-9
assert average_precision_at_3("A", ["B", "C", "D"]) == 0

print("MAP@3 verified ✓")
```

## 6. TF-IDF Baseline

Create a TF-IDF index using the training set's correct answers and rank each option by cosine similarity.

```python
# Build TF-IDF index
correct_texts_train = [
    f"{row['prompt']} {row[row['answer']]}"
    for _, row in train_split.iterrows()
]

tfidf_vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    sublinear_tf=True,
    max_features=80000,
    strip_accents="unicode",
    min_df=1,
)

tfidf_matrix = tfidf_vectorizer.fit_transform(correct_texts_train)

print(f"Vocabulary Size : {len(tfidf_vectorizer.vocabulary_)}")
print(f"Matrix Shape    : {tfidf_matrix.shape}")
```

```python
def tfidf_rank_options(row):
    """Rank answer options using TF-IDF cosine similarity."""

    scores = {}

    for opt in OPTIONS:
        query = f"{row['prompt']} {row[opt]}"
        query_vec = tfidf_vectorizer.transform([query])

        scores[opt] = float(
            cosine_similarity(query_vec, tfidf_matrix).max()
        )

    return sorted(scores, key=lambda x: -scores[x])


# Validation
tfidf_predictions = [
    tfidf_rank_options(row)[:3]
    for _, row in val_split.iterrows()
]

tfidf_map3 = map_at_3(
    val_split["answer"].tolist(),
    tfidf_predictions,
)

top1 = sum(
    pred[0] == ans
    for pred, ans in zip(tfidf_predictions, val_split["answer"])
)

print(f"MAP@3   : {tfidf_map3:.4f}")
print(f"Top-1   : {top1}/{len(val_split)} ({top1/len(val_split):.2%})")
```

## 6b. BM25 Ranking

BM25 is a probabilistic retrieval model that improves on TF-IDF by accounting for document length and term saturation. It provides complementary lexical signal to TF-IDF cosine similarity.


```python
# Build BM25 index over training correct answers
correct_texts_train_raw = [
    str(row["prompt"]) + " " + str(row[row["answer"]])
    for _, row in train_split.iterrows()
]

def tokenize(text):
    return str(text).lower().split()

bm25_corpus   = [tokenize(t) for t in correct_texts_train_raw]
bm25_index    = BM25Okapi(bm25_corpus)
bm25_answers  = [row["answer"] for _, row in train_split.iterrows()]

def bm25_rank_options(row: pd.Series) -> list:
    """Rank answer options using BM25 retrieval scores."""
    scores = {}
    for opt in OPTIONS:
        query  = tokenize(str(row["prompt"]) + " " + str(row[opt]))
        scores[opt] = float(bm25_index.get_scores(query).max())
    return sorted(scores, key=lambda o: -scores[o])

def bm25_score_dict(row: pd.Series) -> dict:
    """Return BM25 scores per option (for soft ensemble)."""
    scores = {}
    for opt in OPTIONS:
        query  = tokenize(str(row["prompt"]) + " " + str(row[opt]))
        scores[opt] = float(bm25_index.get_scores(query).max())
    return scores

print("Evaluating BM25 on validation set...")
bm25_predictions = [bm25_rank_options(row)[:3] for _, row in val_split.iterrows()]
bm25_map3 = map_at_3(val_split["answer"].tolist(), bm25_predictions)
top1_bm25 = sum(p[0] == a for p, a in zip(bm25_predictions, val_split["answer"]))
print(f"MAP@3 : {bm25_map3:.4f}")
print(f"Top-1 : {top1_bm25}/{len(val_split)} ({top1_bm25/len(val_split):.2%})")
```

## 7. Sentence Transformer

Use a pretrained sentence embedding model to compare the semantic similarity between questions and answer options.

```python
def st_rank_options(row: pd.Series) -> list:
    """Rank answer options using sentence embedding similarity."""

    option_texts = [
        f"{row['prompt']} {row[opt]}"
        for opt in OPTIONS
    ]

    option_embs = st_model.encode(
        option_texts,
        convert_to_numpy=True
    )
    option_embs = normalize(option_embs)

    scores = (option_embs @ train_embeddings.T).max(axis=1)

    return [OPTIONS[i] for i in np.argsort(-scores)]


print("Evaluating on validation set...")

st_predictions = [
    st_rank_options(row)[:3]
    for _, row in val_split.iterrows()
]

st_map3 = map_at_3(
    val_split["answer"].tolist(),
    st_predictions
)

top1_st = sum(
    pred[0] == ans
    for pred, ans in zip(st_predictions, val_split["answer"])
)

print(f"MAP@3 : {st_map3:.4f}")
print(f"Top-1 : {top1_st}/{len(val_split)} ({top1_st/len(val_split):.2%})")
```

## 8. Zero-shot NLI

Use a pretrained Natural Language Inference (NLI) model to rank answer options based on how well they match the question.

```python
from transformers import pipeline

print("Loading NLI model...")

# Upgraded from nli-deberta-v3-small → nli-deberta-v3-large
# The large variant has significantly better reasoning and MCQ performance.
nli_pipeline = pipeline(
    "zero-shot-classification",
    model="cross-encoder/nli-deberta-v3-large",
    device=0 if torch.cuda.is_available() else -1,
)

print("Model loaded.")
```

```python
def nli_rank_options(row: pd.Series) -> list:
    """Rank answer options using zero-shot NLI with a task-specific hypothesis."""

    candidate_labels = [row[opt] for opt in OPTIONS]

    result = nli_pipeline(
        row["prompt"],
        candidate_labels=candidate_labels,
        hypothesis_template="The correct answer to this question is: {}",
        multi_label=False,
    )

    label_to_option = {row[opt]: opt for opt in OPTIONS}
    return [label_to_option[label] for label in result["labels"]]


def nli_scores(row: pd.Series) -> dict:
    """Return NLI entailment probability per option (for soft ensemble)."""

    candidate_labels = [row[opt] for opt in OPTIONS]
    result = nli_pipeline(
        row["prompt"],
        candidate_labels=candidate_labels,
        hypothesis_template="The correct answer to this question is: {}",
        multi_label=False,
    )
    label_to_score = dict(zip(result["labels"], result["scores"]))
    return {opt: label_to_score[row[opt]] for opt in OPTIONS}


print("Evaluating NLI on validation set...")
NLI_SAMPLE = len(val_split)
val_sample  = val_split.iloc[:NLI_SAMPLE]

nli_predictions = [nli_rank_options(row)[:3] for _, row in val_sample.iterrows()]
nli_map3 = map_at_3(val_sample["answer"].tolist(), nli_predictions)
top1_nli  = sum(pred[0] == ans for pred, ans in zip(nli_predictions, val_sample["answer"]))

print(f"MAP@3 : {nli_map3:.4f}")
print(f"Top-1 : {top1_nli}/{len(val_sample)} ({top1_nli/len(val_sample):.2%})")
```

## 9. Retrieval-Augmented Ranking (RAG)

Retrieve the most similar training questions and use their correct answers as additional context when ranking the options.

```python
def build_rag_index(train_df: pd.DataFrame, encoder):
    """Build embeddings for training prompts."""

    prompts = train_df["prompt"].tolist()
    correct_answers = [
        row[row["answer"]]
        for _, row in train_df.iterrows()
    ]

    prompt_embs = encoder.encode(
        prompts,
        batch_size=64,
        show_progress_bar=False,
        convert_to_numpy=True,
    )

    prompt_embs = normalize(prompt_embs)

    return prompt_embs, correct_answers


print("Building retrieval index...")

rag_prompt_embs, rag_correct_texts = build_rag_index(
    train_split,
    st_model,
)

print(f"Index Size : {len(rag_prompt_embs)}")
```

```python
def rag_rank_options(row: pd.Series, k: int = 5) -> list:
    """Rank answer options using retrieval-augmented similarity."""
    
    # Retrieve similar questions
    q_emb = st_model.encode([row['prompt']], convert_to_numpy=True)
    q_emb = normalize(q_emb)
    retrieval_scores = (q_emb @ rag_prompt_embs.T)[0]
    top_k_idx = np.argsort(-retrieval_scores)[:k]

    # Build retrieval context
    context_texts = [rag_correct_texts[i] for i in top_k_idx]
    context_embs  = st_model.encode(context_texts, convert_to_numpy=True)
    context_embs  = normalize(context_embs)

    # Score answer options
    option_texts = [f"{row['prompt']} {row[opt]}" for opt in OPTIONS]
    option_embs  = normalize(st_model.encode(option_texts, convert_to_numpy=True))

    sim_corpus  = (option_embs @ train_embeddings.T).max(axis=1)
    sim_context = (option_embs @ context_embs.T).max(axis=1)

    # Combine: give slightly more weight to retrieved context
    combined = 0.6 * sim_corpus + 0.4 * sim_context
    return [OPTIONS[i] for i in np.argsort(-combined)]


print("Evaluating RAG on validation set...")
rag_predictions = [rag_rank_options(row)[:3] for _, row in val_split.iterrows()]
rag_map3 = map_at_3(val_split['answer'].tolist(), rag_predictions)

top1_rag = sum(p[0] == a for p, a in zip(rag_predictions, val_split['answer']))
print(f"MAP@3 : {rag_map3:.4f}")
print(f"Top-1 : {top1_rag}/{len(val_split)} ({top1_rag/len(val_split):.2%})")
```

## 9b. Cross-Encoder Reranking

A cross-encoder directly scores each `(question, option)` pair through a full attention pass, making it much stronger than bi-encoder similarity. It is slower but more accurate.


```python
from sentence_transformers import CrossEncoder

print("Loading cross-encoder...")
# ms-marco-MiniLM-L-12-v2: strong QA relevance scorer, fast on Kaggle T4
cross_encoder = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-12-v2")
print("Cross-encoder loaded.")
```

```python
def ce_rank_options(row: pd.Series) -> list:
    """Rank answer options using cross-encoder relevance scores."""
    pairs  = [(str(row["prompt"]), str(row[opt])) for opt in OPTIONS]
    scores = cross_encoder.predict(pairs)
    return [OPTIONS[i] for i in np.argsort(-scores)]

def ce_score_dict(row: pd.Series) -> dict:
    """Return raw cross-encoder scores per option (for soft ensemble)."""
    pairs  = [(str(row["prompt"]), str(row[opt])) for opt in OPTIONS]
    scores = cross_encoder.predict(pairs)
    return {opt: float(s) for opt, s in zip(OPTIONS, scores)}

print("Evaluating cross-encoder on validation set...")
ce_predictions = [ce_rank_options(row)[:3] for _, row in val_split.iterrows()]
ce_map3  = map_at_3(val_split["answer"].tolist(), ce_predictions)
top1_ce  = sum(p[0] == a for p, a in zip(ce_predictions, val_split["answer"]))
print(f"MAP@3 : {ce_map3:.4f}")
print(f"Top-1 : {top1_ce}/{len(val_split)} ({top1_ce/len(val_split):.2%})")
```

## 10. MLP from Scratch

Train a two-layer MLP using handcrafted TF-IDF similarity features to predict the correct answer option.

```python
def extract_features(df: pd.DataFrame, vec, mat, st_enc=None, st_mat=None) -> tuple:
    """Extract TF-IDF + ST similarity features for the MLP.
    
    Feature vector (25-dim):
      - 5 TF-IDF cosine similarities (one per option)
      - 10 pairwise TF-IDF differences
      - 5 ST cosine similarities (one per option)   ← new
      - 5 BM25 scores (one per option)              ← new
    """
    X, y = [], []
    use_st  = st_enc is not None and st_mat is not None

    for _, row in df.iterrows():
        # TF-IDF sims
        tfidf_sims = []
        for opt in OPTIONS:
            q = str(row["prompt"]) + " " + str(row[opt])
            v = vec.transform([q])
            tfidf_sims.append(float(cosine_similarity(v, mat).max()))

        tfidf_diffs = [tfidf_sims[i] - tfidf_sims[j]
                       for i in range(5) for j in range(i + 1, 5)]

        # ST sims
        if use_st:
            opt_texts = [str(row["prompt"]) + " " + str(row[opt]) for opt in OPTIONS]
            opt_embs  = normalize(st_enc.encode(opt_texts, convert_to_numpy=True, show_progress_bar=False))
            st_sims   = list((opt_embs @ st_mat.T).max(axis=1))
        else:
            st_sims = [0.0] * 5

        # BM25 scores
        bm25_sims = []
        for opt in OPTIONS:
            query = tokenize(str(row["prompt"]) + " " + str(row[opt]))
            bm25_sims.append(float(bm25_index.get_scores(query).max()))

        feats = tfidf_sims + tfidf_diffs + st_sims + bm25_sims
        X.append(feats)
        if "answer" in df.columns:
            y.append(OPTIONS.index(row["answer"]))

    return np.array(X, dtype=np.float32), np.array(y, dtype=np.int32)


print("Extracting features for train split...")
X_train, y_train = extract_features(train_split, tfidf_vectorizer, tfidf_matrix, st_model, train_embeddings)
print("Extracting features for val split...")
X_val,   y_val   = extract_features(val_split,   tfidf_vectorizer, tfidf_matrix, st_model, train_embeddings)

print(f"X_train : {X_train.shape}")
print(f"X_val   : {X_val.shape}")
```

```python
#  MLP in pure NumPy
class MLP:
    """Two-layer MLP implemented in NumPy."""
    
    def __init__(self, n_in, n_hidden, n_out, lr=0.05, seed=42):
        rng = np.random.default_rng(seed)
        scale = np.sqrt(2.0 / n_in)   # He initialisation
        self.W1 = rng.normal(0, scale, (n_in, n_hidden)).astype(np.float32)
        self.b1 = np.zeros(n_hidden, dtype=np.float32)
        self.W2 = rng.normal(0, np.sqrt(2.0 / n_hidden), (n_hidden, n_out)).astype(np.float32)
        self.b2 = np.zeros(n_out, dtype=np.float32)
        self.lr = lr

    def _relu(self, x):
        return np.maximum(0, x)

    def _softmax(self, x):
        ex = np.exp(x - x.max(axis=1, keepdims=True))
        return ex / ex.sum(axis=1, keepdims=True)

    def forward(self, X):
        self.h1  = self._relu(X @ self.W1 + self.b1)
        self.out = self._softmax(self.h1 @ self.W2 + self.b2)
        return self.out

    def backward(self, X, y_true):
        n = len(X)
        # Cross-entropy gradient w.r.t. softmax output
        d_out = self.out.copy()
        d_out[np.arange(n), y_true] -= 1
        d_out /= n
        # Layer 2 gradients
        dW2 = self.h1.T @ d_out
        db2 = d_out.sum(axis=0)
        # Layer 1 gradients (ReLU derivative = 1 if h1 > 0)
        d_h1 = (d_out @ self.W2.T) * (self.h1 > 0)
        dW1  = X.T @ d_h1
        db1  = d_h1.sum(axis=0)
        # Update weights
        self.W1 -= self.lr * dW1
        self.b1 -= self.lr * db1
        self.W2 -= self.lr * dW2
        self.b2 -= self.lr * db2

    def fit(self, X, y, epochs=300, batch_size=128, verbose=True):
        history = []
        n = len(X)
        for epoch in range(1, epochs + 1):
            idx = np.random.permutation(n)
            epoch_loss = 0.0
            for start in range(0, n, batch_size):
                batch = idx[start:start + batch_size]
                xb, yb = X[batch], y[batch]
                probs = self.forward(xb)
                loss  = -np.log(probs[np.arange(len(yb)), yb] + 1e-9).mean()
                epoch_loss += loss * len(yb)
                self.backward(xb, yb)
            avg_loss = epoch_loss / n
            history.append(avg_loss)
            if verbose and epoch % 50 == 0:
                print(f"  Epoch {epoch:3d}/{epochs} — loss: {avg_loss:.4f}")
        return history

    def predict_proba(self, X):
        return self.forward(X)


#  Train 
print("Training MLP from scratch...")
mlp = MLP(n_in=25, n_hidden=128, n_out=5, lr=0.05, seed=SEED)
history = mlp.fit(X_train, y_train, epochs=300, batch_size=128, verbose=True)
```

```python
# Evaluate on validation set 
proba_val = mlp.predict_proba(X_val)
mlp_preds = [[OPTIONS[j] for j in np.argsort(-proba_val[i])][:3]
             for i in range(len(X_val))]
mlp_map3  = map_at_3(val_split['answer'].tolist(), mlp_preds)
top1_mlp  = sum(p[0] == OPTIONS[y_val[i]] for i, p in enumerate(mlp_preds))

print(f"MAP@3 : {mlp_map3:.4f}")
print(f"Top-1 : {top1_mlp}/{len(val_split)} ({top1_mlp/len(val_split):.2%})")

# Log to W&B
wandb.init(
    project=WANDB_PROJECT,
    name="mlp-from-scratch",
    config={
        "model":        "MLP (pure NumPy)",
        "architecture": "15 → 64 → 5",
        "features":     "TF-IDF sim scores + pairwise diffs",
        "activation":   "ReLU + Softmax",
        "optimizer":    "mini-batch SGD",
        "epochs":       300,
        "batch_size":   128,
        "lr":           0.05,
        "val_size":     VAL_SIZE,
    },
    reinit=True
)
for i, loss in enumerate(history, 1):
    wandb.log({"epoch": i, "train_loss": loss})
wandb.log({"val_map3": mlp_map3, "val_top1_acc": top1_mlp / len(val_split)})
wandb.finish()
print("MLP run logged ✓")
```

## 12. Ensemble

Combine predictions from all **6 models** (TF-IDF, Sentence Transformer, NLI, RAG, MLP, LLM)
using **Borda Count voting** with **val-tuned weights** to produce the final top-3 ranked answers.


```python
def softmax(scores: dict) -> dict:
    """Softmax-normalize a score dict over OPTIONS."""
    vals = np.array([scores[o] for o in OPTIONS], dtype=np.float64)
    vals = vals - vals.max()
    exp  = np.exp(vals)
    exp  = exp / exp.sum()
    return {o: float(exp[i]) for i, o in enumerate(OPTIONS)}


def soft_ensemble(score_dicts: list, weights: list) -> list:
    """Weighted average of softmax-normalized score dicts."""
    final = {o: 0.0 for o in OPTIONS}
    for sd, w in zip(score_dicts, weights):
        normed = softmax(sd)
        for o in OPTIONS:
            final[o] += w * normed[o]
    return sorted(final, key=lambda o: -final[o])


# ── Pre-compute all model score dicts on val set ─────────────────────────────
print("Pre-computing scores for all models on val set...")

def tfidf_score_dict(row):
    scores = {}
    for opt in OPTIONS:
        v = tfidf_vectorizer.transform([str(row["prompt"]) + " " + str(row[opt])])
        scores[opt] = float(cosine_similarity(v, tfidf_matrix).max())
    return scores

def st_score_dict(row):
    opt_texts = [str(row["prompt"]) + " " + str(row[opt]) for opt in OPTIONS]
    opt_embs  = normalize(st_model.encode(opt_texts, convert_to_numpy=True, show_progress_bar=False))
    sims      = (opt_embs @ train_embeddings.T).max(axis=1)
    return {opt: float(s) for opt, s in zip(OPTIONS, sims)}

def rag_score_dict(row, k=5):
    q_emb  = normalize(st_model.encode([str(row["prompt"])], convert_to_numpy=True))
    ret_sc = (q_emb @ rag_prompt_embs.T)[0]
    top_k  = np.argsort(-ret_sc)[:k]
    ctx_embs = normalize(st_model.encode(
        [rag_correct_texts[i] for i in top_k], convert_to_numpy=True, show_progress_bar=False))
    opt_texts = [str(row["prompt"]) + " " + str(row[opt]) for opt in OPTIONS]
    opt_embs  = normalize(st_model.encode(opt_texts, convert_to_numpy=True, show_progress_bar=False))
    sim_corp = (opt_embs @ train_embeddings.T).max(axis=1)
    sim_ctx  = (opt_embs @ ctx_embs.T).max(axis=1)
    combined = 0.6 * sim_corp + 0.4 * sim_ctx
    return {opt: float(s) for opt, s in zip(OPTIONS, combined)}

def mlp_score_dict(row):
    sims  = [tfidf_score_dict(row)[o] for o in OPTIONS]
    diffs = [sims[i] - sims[j] for i in range(5) for j in range(i+1, 5)]
    opt_texts = [str(row["prompt"]) + " " + str(row[opt]) for opt in OPTIONS]
    opt_embs  = normalize(st_model.encode(opt_texts, convert_to_numpy=True, show_progress_bar=False))
    st_sims   = list((opt_embs @ train_embeddings.T).max(axis=1))
    bm25_s    = [float(bm25_index.get_scores(tokenize(str(row["prompt"])+" "+str(row[o]))).max()) for o in OPTIONS]
    x = np.array(sims + diffs + st_sims + bm25_s, dtype=np.float32).reshape(1, -1)
    proba = mlp.predict_proba(x)[0]
    return {opt: float(proba[i]) for i, opt in enumerate(OPTIONS)}

print("  TF-IDF...")
pre_tfidf = [tfidf_score_dict(row) for _, row in val_split.iterrows()]
print("  ST...")
pre_st    = [st_score_dict(row)    for _, row in val_split.iterrows()]
print("  NLI...")
pre_nli   = [nli_scores(row)       for _, row in val_split.iterrows()]
print("  RAG...")
pre_rag   = [rag_score_dict(row)   for _, row in val_split.iterrows()]
print("  BM25...")
pre_bm25  = [bm25_score_dict(row)  for _, row in val_split.iterrows()]
print("  Cross-encoder...")
pre_ce    = [ce_score_dict(row)    for _, row in val_split.iterrows()]
print("  MLP...")
pre_mlp   = [mlp_score_dict(row)   for _, row in val_split.iterrows()]

val_answers = val_split["answer"].tolist()

# ── Grid-search weights ───────────────────────────────────────────────────────
print("\nTuning weights on val set...")

best_map3, best_weights = 0.0, (1, 2, 2, 1, 1, 3, 1)

for tw  in [1, 2]:
    for sw  in [2, 3]:
        for nw  in [2, 3]:
            for rw  in [1, 2]:
                for bw  in [1, 2]:
                    for cw  in [2, 3, 4]:
                        for mw  in [1, 2]:
                            preds = []
                            for i in range(len(val_split)):
                                ranked = soft_ensemble(
                                    [pre_tfidf[i], pre_st[i], pre_nli[i],
                                     pre_rag[i], pre_bm25[i], pre_ce[i], pre_mlp[i]],
                                    [tw, sw, nw, rw, bw, cw, mw],
                                )
                                preds.append(ranked[:3])
                            score = map_at_3(val_answers, preds)
                            if score > best_map3:
                                best_map3 = score
                                best_weights = (tw, sw, nw, rw, bw, cw, mw)

tw, sw, nw, rw, bw, cw, mw = best_weights
print(f"\nBest val MAP@3 : {best_map3:.4f}")
print(f"Weights — tfidf:{tw} st:{sw} nli:{nw} rag:{rw} bm25:{bw} ce:{cw} mlp:{mw}")

# ── Evaluate ──────────────────────────────────────────────────────────────────
ens_predictions = [
    soft_ensemble(
        [pre_tfidf[i], pre_st[i], pre_nli[i], pre_rag[i], pre_bm25[i], pre_ce[i], pre_mlp[i]],
        [tw, sw, nw, rw, bw, cw, mw],
    )[:3]
    for i in range(len(val_split))
]

ens_map3 = map_at_3(val_answers, ens_predictions)
top1_ens = sum(p[0] == a for p, a in zip(ens_predictions, val_answers))
print(f"\nEnsemble MAP@3  : {ens_map3:.4f}")
print(f"Top-1           : {top1_ens}/{len(val_split)} ({top1_ens/len(val_split):.2%})")
```

## 14. Generate Submission


```python
print("Training TF-IDF on full data...")
full_correct_texts = [
    str(row["prompt"]) + " " + str(row[row["answer"]])
    for _, row in train_df.iterrows()
]
tfidf_vectorizer_full = TfidfVectorizer(
    ngram_range=(1, 2), sublinear_tf=True,
    max_features=80000, strip_accents="unicode"
)
tfidf_matrix_full = tfidf_vectorizer_full.fit_transform(full_correct_texts)

print("Encoding sentence embeddings on full data (BGE-large)...")
train_embeddings_full = normalize(
    st_model.encode(full_correct_texts, batch_size=32,
                    show_progress_bar=True, convert_to_numpy=True)
)

print("Building RAG index on full data...")
rag_prompt_embs_full  = normalize(
    st_model.encode(train_df["prompt"].tolist(),
                    batch_size=32, show_progress_bar=False, convert_to_numpy=True)
)
rag_correct_texts_full = [str(row[row["answer"]]) for _, row in train_df.iterrows()]

print("Building BM25 index on full data...")
bm25_corpus_full = [tokenize(t) for t in full_correct_texts]
bm25_index_full  = BM25Okapi(bm25_corpus_full)

print("Extracting MLP features on full data...")
X_full, y_full = extract_features(
    train_df, tfidf_vectorizer_full, tfidf_matrix_full,
    st_model, train_embeddings_full
)

print("Training MLP on full data...")
mlp_full = MLP(n_in=25, n_hidden=128, n_out=5, lr=0.05, seed=SEED)
mlp_full.fit(X_full, y_full, epochs=300, batch_size=128, verbose=False)

# NLI pipeline and cross-encoder are stateless — no retraining needed.
print("\nAll models ready for test inference.")
```

```python
def predict_test_row(row: pd.Series) -> list:
    """Predict top-3 using 7-model soft score ensemble with val-tuned weights."""

    prompt = str(row["prompt"])

    # 1. TF-IDF
    tfidf_sc = {}
    for opt in OPTIONS:
        v = tfidf_vectorizer_full.transform([prompt + " " + str(row[opt])])
        tfidf_sc[opt] = float(cosine_similarity(v, tfidf_matrix_full).max())

    # 2. ST (BGE-large)
    opt_texts = [prompt + " " + str(row[opt]) for opt in OPTIONS]
    opt_embs  = normalize(st_model.encode(opt_texts, convert_to_numpy=True, show_progress_bar=False))
    st_sims   = (opt_embs @ train_embeddings_full.T).max(axis=1)
    st_sc     = {opt: float(s) for opt, s in zip(OPTIONS, st_sims)}

    # 3. NLI (DeBERTa-large)
    nli_result = nli_pipeline(
        prompt,
        candidate_labels=[str(row[opt]) for opt in OPTIONS],
        hypothesis_template="The correct answer to this question is: {}",
        multi_label=False,
    )
    label_to_opt = {str(row[opt]): opt for opt in OPTIONS}
    nli_sc = {label_to_opt[lbl]: s for lbl, s in zip(nli_result["labels"], nli_result["scores"])}

    # 4. RAG
    q_emb  = normalize(st_model.encode([prompt], convert_to_numpy=True))
    ret_sc = (q_emb @ rag_prompt_embs_full.T)[0]
    top_k  = np.argsort(-ret_sc)[:5]
    ctx_embs = normalize(st_model.encode(
        [rag_correct_texts_full[i] for i in top_k], convert_to_numpy=True, show_progress_bar=False))
    sim_corp = (opt_embs @ train_embeddings_full.T).max(axis=1)
    sim_ctx  = (opt_embs @ ctx_embs.T).max(axis=1)
    rag_sc   = {opt: float(0.6 * sim_corp[i] + 0.4 * sim_ctx[i]) for i, opt in enumerate(OPTIONS)}

    # 5. BM25
    bm25_sc = {}
    for opt in OPTIONS:
        q = tokenize(prompt + " " + str(row[opt]))
        bm25_sc[opt] = float(bm25_index_full.get_scores(q).max())

    # 6. Cross-encoder
    pairs = [(prompt, str(row[opt])) for opt in OPTIONS]
    ce_scores = cross_encoder.predict(pairs)
    ce_sc = {opt: float(s) for opt, s in zip(OPTIONS, ce_scores)}

    # 7. MLP
    sims   = [tfidf_sc[o] for o in OPTIONS]
    diffs  = [sims[i] - sims[j] for i in range(5) for j in range(i+1, 5)]
    bm25_f = [bm25_sc[o] for o in OPTIONS]
    x      = np.array(sims + diffs + list(st_sims) + bm25_f, dtype=np.float32).reshape(1, -1)
    proba  = mlp_full.predict_proba(x)[0]
    mlp_sc = {opt: float(proba[i]) for i, opt in enumerate(OPTIONS)}

    return soft_ensemble(
        [tfidf_sc, st_sc, nli_sc, rag_sc, bm25_sc, ce_sc, mlp_sc],
        [tw, sw, nw, rw, bw, cw, mw],
    )[:3]


print("Generating submission...")
rows = []
for i, (_, row) in enumerate(test_df_clean.iterrows()):
    pred = predict_test_row(row)
    rows.append({"ID": row["id"], "Prediction": " ".join(pred)})
    if (i + 1) % 50 == 0:
        print(f"  {i+1}/{len(test_df_clean)}")

submission = pd.DataFrame(rows)
submission.to_csv(OUTPUT_PATH, index=False)
print(f"\nSaved to {OUTPUT_PATH}")
print(submission.head(10).to_string(index=False))
```
